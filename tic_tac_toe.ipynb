```python
import random

# Display the board
def display_board(board):
    print("\n")
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("---+---+---")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("---+---+---")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print("\n")


# Check whether a player has won
def check_winner(board, player):
    winning_combinations = [
        [0, 1, 2],  # Row 1
        [3, 4, 5],  # Row 2
        [6, 7, 8],  # Row 3
        [0, 3, 6],  # Column 1
        [1, 4, 7],  # Column 2
        [2, 5, 8],  # Column 3
        [0, 4, 8],  # Diagonal
        [2, 4, 6]   # Diagonal
    ]

    for combination in winning_combinations:
        if (board[combination[0]] == player and
            board[combination[1]] == player and
            board[combination[2]] == player):
            return True

    return False


# Check whether the board is full
def check_draw(board):
    return " " not in board


# Main game
def play_game():
    board = [" "] * 9

    human = "X"
    computer = "O"

    print("TIC-TAC-TOE")
    print("You are X and Computer is O.")
    print("Enter a number from 1 to 9 to place your X.")

    # Show position numbers
    print("\nBoard positions:")
    print(" 1 | 2 | 3")
    print("---+---+---")
    print(" 4 | 5 | 6")
    print("---+---+---")
    print(" 7 | 8 | 9")

    while True:
        # HUMAN TURN
        display_board(board)

        while True:
            try:
                position = int(input("Enter your position (1-9): "))

                if position < 1 or position > 9:
                    print("Please enter a number between 1 and 9.")
                elif board[position - 1] != " ":
                    print("That position is already occupied.")
                else:
                    break

            except ValueError:
                print("Please enter a valid number.")

        board[position - 1] = human

        # Check if human wins
        if check_winner(board, human):
            display_board(board)
            print("Congratulations! You WIN!")
            break

        # Check for draw
        if check_draw(board):
            display_board(board)
            print("The game is a DRAW!")
            break

        # COMPUTER TURN
        available_positions = [
            i for i in range(9) if board[i] == " "
        ]

        computer_position = random.choice(available_positions)
        board[computer_position] = computer

        print("Computer chose position:", computer_position + 1)

        # Check if computer wins
        if check_winner(board, computer):
            display_board(board)
            print("Computer WINS!")
            break

        # Check for draw
        if check_draw(board):
            display_board(board)
            print("The game is a DRAW!")
            break


# Start the game
play_game()
```
